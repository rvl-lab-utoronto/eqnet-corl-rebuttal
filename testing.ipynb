```python
import numpy as np

def combine_mean_std(means, stds, n):
    """
    Recover the mean and std of the pooled sample from per-group
    means and stds (as computed by numpy.std, ddof=0), assuming
    each group has the same size n.

    means, stds : array-like, one entry per group
    n           : sample size of each group (same for all groups)
    """
    means = np.asarray(means, dtype=float)
    stds = np.asarray(stds, dtype=float)
    k = len(means)  # number of groups

    grand_mean = means.mean()

    within_group_var = np.mean(stds**2)
    between_group_var = np.mean((means - grand_mean)**2)

    total_var = within_group_var + between_group_var
    total_std = np.sqrt(total_var)

    return grand_mean, total_std


# Example usage
means = [0.64467684, 0.63285696, 0.71146125]
stds = [0.20691928, 0.20664881, 0.17890446]
n = 150  # size of each group

grand_mean, total_std = combine_mean_std(means, stds, n)
print(grand_mean*100, total_std*100)
```
